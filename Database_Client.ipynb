```python
import mysql.connector as connector
```

```python
connection = connector.connect(user = "your_user", password = "your_password", db = "littlelemondb")
print("Connection to the database successfully established")
```

```python
# Create the cursor object
cursor = connection.cursor()

print("Cursor instance successfully initiated!")
```

```python
show_tables_query = "SHOW tables" 
cursor.execute(show_tables_query)
```

```python
# Fetch all the results from the cursor
results = cursor.fetchall()

# Print the table names
print("Tables in the Little Lemon database:")
for table in results:
    print(table[0])
```

QUERY WITH TABLE JOIN


```python
join_tables_query = """SELECT CONCAT(c.CustomerFirstName, ' ', c.CustomerLastName) 
AS " Customer Full Name",
c.ContactNumber, c.Email, o.TotalCost 
FROM CustomerDetails AS c INNER JOIN Bookings AS b ON c.CustomerID = b.CustomerID
INNER JOIN Orders AS o ON b.BookingID = o.BookingID
WHERE o.TotalCost > 60"""
```

```python
#Execute the join query
cursor.execute(join_tables_query)

#Fetch the results
results = cursor.fetchall()

# Print the header 
print(f"{'Full Name':<25} | {'Phone':<15} | {'Email':<25} | {'Cost':<10}")
print("-" * 80)

# Iterate and print the rows
for row in results:
    name, phone, email, cost = row
    print(f"{name:<25} | {phone:<15} | {email:<25} | ${cost:<10}")
```

```python
# Check if the connection is still open before trying to close it
if connection.is_connected():
    cursor.close()
    connection.close()
    print("MySQL connection is now closed.")
```
